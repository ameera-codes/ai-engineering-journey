```python
print("wake up")
print("start machine")
print("drink coffee")
print("enjoy coffee")
print("eat breakfast")
print("go to work")
print("come home")
```

```python
def make_coffee():
    print("start machine")
    print("drink coffee")
    print("enjoy coffee")
```

```python
print("wake up")
make_coffee()
print("eat breakfast")
print("go to work")
print("come home")
```

```python
def daily_routine():
    print("wake up")
    make_coffee()
    print("eat breakfast")
    print("go to work")
    print("come home")
```

```python
daily_routine()
```
