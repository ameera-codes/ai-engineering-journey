```python
#parameters and arguments
def clean_name(name):
    print(name.strip().lower())
    #strip() removes whitespace from the beginning and end of the string,
    #lower() converts the string to lowercase

clean_name("  Alice  ") #function call with argument
```

```python
case_rule = "lower" #global variable

def clean_name(name): #paramenter
    cleaned = name.strip().lower() #local variable
    if case_rule == "lower":
        cleaned = cleaned.lower()
    print("Cleaned:", cleaned)

clean_name("  Alice  ") #function call with argument
print("The case rule is:", case_rule) #accessing global variable
```

```python
def clean_name(first_name, last_name, country):
    first = first_name.strip().lower() 
    last = last_name.strip().lower()
    country = country.strip().lower()
    
    print(first, " ", last, " ", country) #positional arguments

clean_name("  Alice  ", "  Smith  ", "  USA  ") #function call with positional arguments

clean_name(country="  USA  ", last_name="  Smith  ", first_name="  Alice  ") #function call with keyword arguments
```
