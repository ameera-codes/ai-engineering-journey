```python
import math #math is a built-in module in Python that provides mathematical functions

result = math.sqrt(16) #sqrt calculates the square root and accessing the sqrt function from the math module

print(result)
```

```python
from math import sqrt #importing only the sqrt function from the math module

result = sqrt(16) #using the sqrt function directly
print(result)
```

```python
import math as m #importing the math module with an alias 'm'

result = m.sqrt(16) #using the alias 'm' to access the sqrt function
print(result)
```
